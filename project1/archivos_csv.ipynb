```python
# create 4 years of data
full_df = df.copy()
days = 245
for i in range(6):
    df['Date'] = pd.to_datetime(df['Date']) - pd.DateOffset(days=days)
    full_df = pd.concat([full_df, df], axis=0)
```

```python
days = 300
df['Date'] = pd.to_datetime(df['Date']) + pd.DateOffset(days=days)
full_df = pd.concat([full_df, df], axis=0)
```

```python
print(max(pd.to_datetime(full_df.Date)) - min(pd.to_datetime(full_df.Date)))
print(full_df.shape)
```

```python
# crear archivos por cada año mes de la data en full_df
for i in range(2018, 2026):
    for j in range(1, 13):
        try:
            df = full_df[(pd.to_datetime(full_df.Date).dt.year == i) & (pd.to_datetime(full_df.Date).dt.month == j)]
            df.to_csv(f'data/car_sales_data_{i}_{j}.csv', index=False)
            print(f'data/car_sales_data_{i}_{j}.csv')
        except:
            print(f'No data for {i} - {j}')
```
